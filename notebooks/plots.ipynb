```python
import patsy              # already bundled with statsmodels
import statsmodels.api as sm
from itertools import chain

def run_linear_inference(df, mode, y_col: str = "Edit_Distance"):
    """
    Fit an OLS with all main effects + two-way interactions and return the statsmodels results object.
    """
    # 1) Sub-set if necessary
    if mode == "frq":
        formula = (
            f"{y_col} ~ Length + Zipf_Frequency + C(Morphology)"
            " + Length:C(Morphology) + Zipf_Frequency:C(Morphology)"
        )
    elif mode == "lex":
        formula = (
            f"{y_col} ~ Length + C(Lexicality) + C(Morphology)"
            " + Length:C(Lexicality) + C(Lexicality):C(Morphology) + Length:C(Morphology)"
        )

    # 2) Build design matrices & fit
    y, X = patsy.dmatrices(formula, data=df, return_type="dataframe")
    model = sm.OLS(y, X).fit(cov_type="HC3")   # HC3 = robust (White) SEs
    return model
```

```python
# get all files in sibling "ablations" directory

ablations_dir = os.path.join(parent, "notebooks", "ablations")
files = [os.path.join(ablations_dir, f) for f in os.listdir(ablations_dir) if f.endswith(".csv")]

# for each file, read it, run inference, and print results
for f in files:
    df = pd.read_csv(f)
    # df["Zipf_Frequency"] = df["Zipf_Frequency"].fillna(0)
    df = df[df["Lexicality"] == "real"]

    res = run_linear_inference(df, mode="frq", y_col="Edit_Distance")
    # print(f"\n{os.path.basename(f)}:")
    # print(res.pvalues)

    def main_effect(name):
        """Return the first p-value row that matches `name` but has no ':'."""
        rows = [k for k in res.pvalues.index if k.startswith(name) and ":" not in k]
        return res.pvalues[rows[0]] if rows else None

    p_zipf   = main_effect("Zipf_Frequency")
    p_length = main_effect("Length")

    # ---- report only when Zipf_Frequency is below threshold -----------------
    if p_zipf is not None and p_zipf <= 0.005:
        print(
            f"{os.path.basename(f)}\t"
            f"p(Zipf_Frequency) = {p_zipf:.4g}\t"
            f"p(Length) = {p_length:.4g}" if p_length is not None else "p(Length) = N/A"
        )
```

```python
from swp.viz.embeddings import find_best_n_clusters

df = pd.read_csv("data/mlem_univariate.csv", index_col=0)
df = df[(df["Importance"] > 0) & (df['Spearman'] > 0.003)]
# df["Importance"] /= np.sqrt(df["Spearman"])
df["Importance"] /= df["Spearman"]

pivot = df.copy()
pivot = pivot.pivot_table(index='Neuron', columns='Feature', values='Importance')
pivot = pivot.fillna(0)

best_num_clusters, clusters = find_best_n_clusters(pivot, min_clusters=2)
print(f"Optimal number of clusters: {best_num_clusters}")
pivot["Cluster"] = clusters
pivot = pivot.reset_index().rename(columns={'index': 'Neuron'})
df = df.merge(pivot[["Neuron", "Cluster"]], on="Neuron")
```

```python
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import seaborn as sns

df_m = df.copy()

# Compute cumulative sums per neuron for stacking
df_m = df_m.sort_values(['Neuron', 'Importance'], ascending=[True, False])
df_m['cum_importance'] = df_m.groupby('Neuron')['Importance'].cumsum()
df_m['bottom'] = df_m['cum_importance'] - df_m['Importance']

# Compute total importance per neuron.
neuron_totals = df_m.groupby('Neuron')['Importance'].sum().reset_index().rename(
    columns={'Importance': 'total_importance'}
)

# Get unique neuron-level info (Neuron and Cluster) and merge totals.
neurons = df_m[['Neuron', 'Cluster']].drop_duplicates().reset_index(drop=True)
neurons = neurons.merge(neuron_totals, on="Neuron", how="left")

# Compute total importance per cluster and sort clusters descending.
cluster_totals = neurons.groupby("Cluster")["total_importance"].sum().reset_index()
cluster_totals = cluster_totals.sort_values("total_importance", ascending=False).reset_index(drop=True)

# Create a mapping from the original Cluster to a new cluster order
cluster_order_map = {old: new for new, old in enumerate(cluster_totals["Cluster"], start=1)}
neurons["new_cluster"] = neurons["Cluster"].map(cluster_order_map)

# Sort neurons by new cluster order and then by total importance (or neuron id)
neurons = neurons.sort_values(["new_cluster", "total_importance"], ascending=[True, False]).reset_index(drop=True)
neurons["within_cluster_order"] = neurons.groupby("new_cluster").cumcount()

# Define a custom x-axis (unit_index) with gaps between clusters.
gap_size = 5  # adjust gap size as needed
offset_map = {}
current_offset = 0
for cl in sorted(neurons["new_cluster"].unique()):
    offset_map[cl] = current_offset
    count_in_cluster = neurons[neurons["new_cluster"] == cl].shape[0]
    current_offset += count_in_cluster + gap_size

neurons["unit_index"] = neurons["within_cluster_order"] + neurons["new_cluster"].map(offset_map)

# Merge the custom unit_index back into the melted DataFrame.
df_m = df_m.merge(neurons[["Neuron", "unit_index"]], on="Neuron", how="left")

# Plot the stacked bars.
plt.figure(figsize=(22, 9))
ax = plt.gca()

# Use Seaborn's default "deep" palette for our features.
features = sorted(df_m["Feature"].unique())
palette = dict(zip(features, sns.color_palette("deep", len(features))))

# Draw each segment as a bar segment.
for _, row in df_m.iterrows():
    ax.bar(
        row['unit_index'],
        row['Importance'],
        bottom=row['bottom'],
        width=1,
        color=palette.get(row['Feature'], 'gray')
    )

# ax.set_xlabel("Neuron (with clusters sorted by total FI & gaps between clusters)")
ax.set_ylabel("Stacked Feature Importance", fontsize=20)
# ax.set_title("Stacked Feature Importance per Neuron")
plt.xticks([])  # Optionally hide x-tick labels

handles = [
    mlines.Line2D([], [], color=palette[feat], marker='s', linestyle='None', markersize=10, label=feat)
    for feat in features
]
plt.legend(handles=handles, title="Feature", fontsize=18, title_fontsize=20)
plt.show()
```
